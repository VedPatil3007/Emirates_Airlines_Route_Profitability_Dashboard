```python
queries = {

    "Q1. Total flights, passengers, revenue, cost and profit": """
        SELECT
            COUNT(*) AS total_flights,
            SUM(passengers) AS total_passengers,
            SUM(total_revenue) AS total_revenue,
            SUM(total_cost) AS total_cost,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability;
    """,

    "Q2. Overall average load factor and profit margin": """
        SELECT
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor,
            ROUND(AVG(profit_margin)::numeric, 2) AS avg_profit_margin
        FROM emirates_airlines_route_profitability;
    """,

    "Q3. Business performance across different seasons": """
        SELECT
            season,
            COUNT(*) AS total_flights,
            SUM(passengers) AS total_passengers,
            SUM(total_revenue) AS total_revenue,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY season
        ORDER BY total_revenue DESC;
    """,

    "Q4. Revenue and profit by demand category": """
        SELECT
            demand_level,
            COUNT(*) AS total_flights,
            SUM(total_revenue) AS total_revenue,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY demand_level
        ORDER BY total_revenue DESC;
    """,

    "Q5. Monthly revenue and profit trends": """
        SELECT
            DATE_TRUNC('month', TO_DATE(flight_date, 'DD-MM-YYYY')) AS flight_month,
            SUM(total_revenue) AS total_revenue,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY flight_month
        ORDER BY flight_month;
    """,

    "Q6. Highest and lowest profit-generating months": """
        SELECT
            EXTRACT(MONTH FROM TO_DATE(flight_date, 'DD-MM-YYYY')) AS month_number,
            TO_CHAR(TO_DATE(flight_date, 'DD-MM-YYYY'), 'Month') AS month,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY
            EXTRACT(MONTH FROM TO_DATE(flight_date, 'DD-MM-YYYY')),
            TO_CHAR(TO_DATE(flight_date, 'DD-MM-YYYY'), 'Month')
        ORDER BY total_profit DESC;
    """,

    "Q7. Profitability across route categories": """
        SELECT
            route_category,
            COUNT(*) AS total_flights,
            SUM(total_revenue) AS total_revenue,
            SUM(profit) AS total_profit,
            ROUND(AVG(profit_margin)::numeric, 2) AS avg_profit_margin
        FROM emirates_airlines_route_profitability
        GROUP BY route_category
        ORDER BY total_profit DESC;
    """,

    "Q8. Average load factor and profit margin by demand level": """
        SELECT
            demand_level,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor,
            ROUND(AVG(profit_margin)::numeric, 2) AS avg_profit_margin
        FROM emirates_airlines_route_profitability
        GROUP BY demand_level
        ORDER BY avg_profit_margin DESC;
    """,

    "Q9. Top 10 most profitable flight routes": """
        SELECT
            route,
            COUNT(*) AS total_flights,
            SUM(total_revenue) AS total_revenue,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY route
        ORDER BY total_profit DESC
        LIMIT 10;
    """,

    "Q10. Top 10 revenue-generating routes": """
        SELECT
            route,
            SUM(total_revenue) AS total_revenue,
            SUM(passengers) AS total_passengers
        FROM emirates_airlines_route_profitability
        GROUP BY route
        ORDER BY total_revenue DESC
        LIMIT 10;
    """,

    "Q11. Routes operating at a financial loss": """
        SELECT
            route,
            COUNT(*) AS total_flights,
            SUM(total_revenue) AS total_revenue,
            SUM(total_cost) AS total_cost,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY route
        HAVING SUM(profit) < 0
        ORDER BY total_profit ASC;
    """,

    "Q12. Origin and destination combinations with highest load factor": """
        SELECT
            origin,
            destination,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor,
            COUNT(*) AS total_flights
        FROM emirates_airlines_route_profitability
        GROUP BY origin, destination
        ORDER BY avg_load_factor DESC
        LIMIT 10;
    """,

    "Q13. Revenue contribution by source": """
        SELECT
            SUM(ticket_revenue) AS ticket_revenue,
            SUM(ancillary_revenue) AS ancillary_revenue,
            SUM(total_revenue) AS total_revenue
        FROM emirates_airlines_route_profitability;
    """,

    "Q14. Cost components contributing to total expenditure": """
        SELECT 'Fuel Cost' AS cost_component, SUM(fuel_cost) AS total_cost
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Maintenance Cost', SUM(maintenance_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Crew Cost', SUM(crew_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Depreciation Cost', SUM(depreciation_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Insurance Cost', SUM(insurance_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Airport Fees', SUM(airport_fees)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Catering Cost', SUM(catering_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Handling Cost', SUM(handling_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Navigation Fees', SUM(navigation_fees)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Sales Distribution Cost', SUM(sales_distribution_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Passenger Service Cost', SUM(passenger_service_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Overhead Cost', SUM(overhead_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'Marketing Cost', SUM(marketing_cost)
        FROM emirates_airlines_route_profitability

        UNION ALL

        SELECT 'IT Systems Cost', SUM(it_systems_cost)
        FROM emirates_airlines_route_profitability

        ORDER BY total_cost DESC;
    """,

    "Q15. Routes with highest profit margins": """
        SELECT
            route,
            ROUND(AVG(profit_margin)::numeric, 2) AS avg_profit_margin,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY route
        ORDER BY avg_profit_margin DESC
        LIMIT 10;
    """,

    "Q16. Profitable versus loss-making flights": """
        SELECT
            CASE
                WHEN profit > 0 THEN 'Profitable'
                WHEN profit < 0 THEN 'Loss Making'
                ELSE 'Break Even'
            END AS flight_status,
            COUNT(*) AS total_flights,
            SUM(profit) AS total_profit
        FROM emirates_airlines_route_profitability
        GROUP BY flight_status;
    """,

    "Q17. Monthly passenger traffic trends": """
        SELECT
            DATE_TRUNC('month', TO_DATE(flight_date, 'DD-MM-YYYY')) AS flight_month,
            SUM(passengers) AS total_passengers,
            COUNT(*) AS total_flights,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor
        FROM emirates_airlines_route_profitability
        GROUP BY flight_month
        ORDER BY flight_month;
    """,

    "Q18. Top 10 routes by passenger traffic": """
        SELECT
            route,
            SUM(passengers) AS total_passengers,
            COUNT(*) AS total_flights,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor
        FROM emirates_airlines_route_profitability
        GROUP BY route
        ORDER BY total_passengers DESC
        LIMIT 10;
    """,

    "Q19. Passenger demand across different seasons": """
        SELECT
            season,
            SUM(passengers) AS total_passengers,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor,
            COUNT(*) AS total_flights
        FROM emirates_airlines_route_profitability
        GROUP BY season
        ORDER BY total_passengers DESC;
    """,

    "Q20. Passenger traffic by demand category": """
        SELECT
            demand_level,
            SUM(passengers) AS total_passengers,
            COUNT(*) AS total_flights,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor
        FROM emirates_airlines_route_profitability
        GROUP BY demand_level
        ORDER BY total_passengers DESC;
    """,

    "Q21. Passenger and capacity analysis by aircraft type": """
        SELECT
            aircraft_type,
            COUNT(*) AS total_flights,
            SUM(passengers) AS total_passengers,
            SUM(aircraft_capacity) AS total_capacity,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor
        FROM emirates_airlines_route_profitability
        GROUP BY aircraft_type
        ORDER BY total_passengers DESC;
    """,

    "Q22. Average profit by aircraft type": """
        SELECT
            aircraft_type,
            COUNT(*) AS total_flights,
            ROUND(AVG(profit)::numeric, 2) AS avg_profit_per_flight,
            ROUND(AVG(profit_margin)::numeric, 2) AS avg_profit_margin
        FROM emirates_airlines_route_profitability
        GROUP BY aircraft_type
        ORDER BY avg_profit_per_flight DESC;
    """,

    "Q23. Average flight duration by aircraft type": """
        SELECT
            aircraft_type,
            ROUND(AVG(flight_hours)::numeric, 2) AS avg_flight_hours,
            SUM(flight_hours) AS total_flight_hours,
            COUNT(*) AS total_flights
        FROM emirates_airlines_route_profitability
        GROUP BY aircraft_type
        ORDER BY total_flight_hours DESC;
    """,

    "Q24. Flight duration and passenger utilization by route": """
        SELECT
            route,
            ROUND(AVG(flight_hours)::numeric, 2) AS avg_flight_hours,
            ROUND((AVG(load_factor) * 100)::numeric, 2) AS avg_load_factor,
            SUM(passengers) AS total_passengers
        FROM emirates_airlines_route_profitability
        GROUP BY route
        ORDER BY avg_flight_hours DESC
        LIMIT 10;
    """
}
```

```python
import pandas as pd

results = {}

for question, query in queries.items():
    results[question] = pd.read_sql_query(query, engine)

print("All queries executed successfully!")
```

```python
from reportlab.lib import colors
from reportlab.lib.pagesizes import A4, landscape
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import mm
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer,
    Table, TableStyle, Preformatted, KeepTogether
)
from reportlab.lib.colors import HexColor
from xml.sax.saxutils import escape
import pandas as pd
import os

pdf_file = "Emirates_Airlines_SQL_Analysis.pdf"

doc = SimpleDocTemplate(
    pdf_file,
    pagesize=landscape(A4),
    rightMargin=15*mm,
    leftMargin=15*mm,
    topMargin=15*mm,
    bottomMargin=15*mm
)

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    "TitleCustom",
    parent=styles["Title"],
    fontSize=20,
    alignment=1,
    textColor=HexColor("#263B5A"),
    spaceAfter=10
)

subtitle_style = ParagraphStyle(
    "Subtitle",
    parent=styles["Normal"],
    fontSize=11,
    alignment=1,
    spaceAfter=20
)

question_style = ParagraphStyle(
    "QuestionCustom",
    parent=styles["Heading2"],
    fontSize=11,
    leading=14,
    textColor=HexColor("#B22222"),
    spaceBefore=12,
    spaceAfter=8
)

sql_style = ParagraphStyle(
    "SQLCustom",
    fontName="Courier",
    fontSize=7,
    leading=9
)

elements = []

elements.append(Paragraph(
    "EMIRATES AIRLINES",
    title_style
))

elements.append(Paragraph(
    "SQL Business Analysis – Questions, Queries and Outputs",
    subtitle_style
))

elements.append(Spacer(1, 10))

for question, query in queries.items():

    df_result = results[question]

    elements.append(Paragraph(
        escape(question),
        question_style
    ))

    elements.append(Paragraph(
        "<b>SQL Query:</b>",
        styles["Normal"]
    ))

    elements.append(Spacer(1, 5))

    elements.append(Preformatted(
        query.strip(),
        sql_style
    ))

    elements.append(Spacer(1, 8))

    elements.append(Paragraph(
        "<b>Query Output:</b>",
        styles["Normal"]
    ))

    elements.append(Spacer(1, 5))

    if df_result.empty:

        elements.append(Paragraph(
            "No records returned.",
            styles["Normal"]
        ))

    else:

        data = [
            [str(column) for column in df_result.columns]
        ]

        for row in df_result.itertuples(index=False, name=None):
            data.append([
                str(value) if pd.notna(value) else "NULL"
                for value in row
            ])

        table = Table(
            data,
            repeatRows=1,
            hAlign="LEFT"
        )

        table.setStyle(TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), HexColor("#263B5A")),
            ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
            ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
            ("FONTNAME", (0, 1), (-1, -1), "Helvetica"),
            ("FONTSIZE", (0, 0), (-1, -1), 8),
            ("GRID", (0, 0), (-1, -1), 0.4, colors.grey),
            ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
            ("ROWBACKGROUNDS", (0, 1), (-1, -1),
             [colors.white, HexColor("#EDF2F7")]),
            ("LEFTPADDING", (0, 0), (-1, -1), 6),
            ("RIGHTPADDING", (0, 0), (-1, -1), 6),
            ("TOPPADDING", (0, 0), (-1, -1), 5),
            ("BOTTOMPADDING", (0, 0), (-1, -1), 5)
        ]))

        elements.append(table)

    elements.append(Spacer(1, 18))

doc.build(elements)

print("PDF generated successfully!")
print("Location:", os.path.abspath(pdf_file))
```
